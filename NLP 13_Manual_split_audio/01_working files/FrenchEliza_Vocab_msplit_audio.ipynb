```python
output_names: dict[int|float, str] = {
    # Mar, 31, 2026

    1:  "fr_Elisa02_001_Can Elisa hear us.wav"
    ,10:  "fr_Elisa02_010_Some were didn't agree at all and said that anyway.wav"
    ,11:  "fr_Elisa02_011_the French are don't understand on purpose when you talk to them.wav"
    ,12:  "fr_Elisa02_012_and they are simply arrogant.wav"

    }

# added _manual automatically at the end of the file name
import re
def add_manual(fname: str) -> str:
    low = fname.lower()
    if low.endswith('_manual.wav') or low.endswith('_manual.mp3'):
        return fname
    return re.sub(r'(\.(?:wav|mp3))$', r'_manual\1', fname, flags=re.IGNORECASE)

output_names_adj: dict[int|float, str] = {k: add_manual(v) for k, v in output_names.items()}
```
